```python
nu = ['int16', 'int32','int64', 'float16', 'float32', 'float64']    # selecting only numerical features from the dataset
nu_features = list(data.select_dtypes(include=nu).columns)
```

```python
outlier = data[nu_features]
```

```python
sorted(outlier)
```

```python
Q1 = outlier.quantile(0.25)
Q3 = outlier.quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

```python
plt.scatter(data['Age'], data['ChestPainType'], color='red')
plt.title('ChestPainType Vs Age', fontsize=14)
plt.xlabel('Age', fontsize=14)
plt.ylabel('ChestPainType', fontsize=14)
plt.grid(True)
plt.show()
```

```python
from sklearn.impute import SimpleImputer # Importing the SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
outlier_out = outlier[~((outlier < (Q1 - 1.5 * IQR)) |(outlier > (Q3 + 1.5 * IQR))).any(axis = 1)]
print(outlier_out.shape)
```

```python
imputer = SimpleImputer(strategy='mean')
data_imputed = imputer.fit_transform(outlier_out)
# Scaling Data
scaler = StandardScaler()
data_scaled = scaler.fit_transform(outlier_out)
print("Data Preprocessing Completed")
```
