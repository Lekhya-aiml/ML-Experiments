```python
import pandas as pd
import numpy as np

from sklearn.naive_bayes import CategoricalNB
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

data = pd.read_csv("/content/diabetes (1).csv")

print("Dataset:\n", data.head())


encoders = {}

for column in data.columns:
    le = LabelEncoder()
    data[column] = le.fit_transform(data[column])
    encoders[column] = le

print("\nEncoded Data:\n", data.head())


X = data.iloc[:, :-1]
y = data.iloc[:, -1]

print("\nFeatures:\n", X.head())
print("\nTarget:\n", y.head())


X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

n_categories = [X[col].nunique() for col in X.columns]


model = CategoricalNB(min_categories=n_categories)
model.fit(X_train, y_train)

print("\nModel trained successfully!")


y_pred = model.predict(X_test)

print("\nPredictions:\n", y_pred)


accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)
print("\nClassification Report:\n", classification_report(y_test, y_pred))


new_sample = {
    "Pregnancies": 2,
    "Glucose": 120,
    "BloodPressure": 70,
    "SkinThickness": 30,
    "Insulin": 100,
    "BMI": 30.0,
    "DiabetesPedigreeFunction": 0.627,
    "Age": 30
}

new_df = pd.DataFrame([new_sample])

new_df = new_df[X_train.columns]

for column in new_df.columns:
    if column in encoders:
        new_df[column] = encoders[column].transform(new_df[column])
    else:
        print(f"Warning: No encoder found for column '{column}'. Skipping encoding.")

prediction = model.predict(new_df)

predicted_label = encoders[data.columns[-1]].inverse_transform(prediction)

print("\nNew Sample Prediction:", predicted_label[0])
```
